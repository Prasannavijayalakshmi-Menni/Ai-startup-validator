```python

%%writefile tavily_search.py
import os
import json
from tavily import TavilyClient


def search_startup_idea(startup_idea, client):
    if not startup_idea or not startup_idea.strip():
        raise ValueError("Please enter a startup idea.")

    queries = {
        "Market Demand": (
            f'"{startup_idea}" market demand trends target customers India'
        ),
        "Competitor Analysis": (
            f'"{startup_idea}" competitors alternatives similar products India'
        )
    }

    all_results = {}

    for category, query in queries.items():
        print(f"\nSearching: {category}")
        print("Query:", query)

        try:
            response = client.search(
                query=query,
                max_results=5
            )
            all_results[category] = response.get("results", [])

        except Exception as error:
            print(f"Search failed for {category}: {error}")
            all_results[category] = []

    return all_results


def main():
    api_key = os.getenv("TAVILY_API_KEY")

    if not api_key:
        print("TAVILY_API_KEY is missing.")
        print("Set the API key as an environment variable before running.")
        return

    startup_idea = input("Enter your startup idea: ").strip()

    if not startup_idea:
        print("Startup idea cannot be empty.")
        return

    client = TavilyClient(api_key=api_key)
    results = search_startup_idea(startup_idea, client)

    for category, items in results.items():
        print(f"\n--- {category}: {len(items)} results ---")

        for number, item in enumerate(items, start=1):
            print(f"\n{number}. {item.get('title', 'No title')}")
            print("URL:", item.get("url", "No URL"))
            print("Summary:", item.get("content", "No summary"))

    with open("tavily_search_results.json", "w", encoding="utf-8") as file:
        json.dump(results, file, indent=2, ensure_ascii=False)

    print("\nResults saved to tavily_search_results.json")


if __name__ == "__main__":
    main()
```

```python

%%writefile tavily_search.py

import os
import json
from tavily import TavilyClient


def search_startup_idea(startup_idea, client):
    """Search for market demand and competitors for a startup idea."""

    if not startup_idea or not startup_idea.strip():
        raise ValueError("Please enter a startup idea.")

    queries = {
        "Market Demand": (
            f"{startup_idea} market demand trends target customers India"
        ),
        "Competitor Analysis": (
            f"{startup_idea} competitors alternatives similar products India"
        ),
    }

    all_results = {}

    for category, query in queries.items():
        print(f"\nSearching: {category}")
        print("Search query:", query)

        try:
            response = client.search(query=query, max_results=5)
            results = response.get("results", [])
            all_results[category] = results

            for number, result in enumerate(results, start=1):
                print(f"\n--- {category}: Result {number} ---")
                print("Title:", result.get("title", "No title"))
                print("URL:", result.get("url", "No URL"))
                print("Summary:", result.get("content", "No summary"))

        except Exception as error:
            print(f"Search failed for {category}: {error}")
            all_results[category] = []

    return all_results


def main():
    api_key = os.getenv("TAVILY_API_KEY")

    if not api_key:
        print("TAVILY_API_KEY is missing.")
        print("Please configure it as an environment variable.")
        return

    startup_idea = input("Enter your startup idea: ").strip()

    if not startup_idea:
        print("Startup idea cannot be empty.")
        return

    client = TavilyClient(api_key=api_key)
    results = search_startup_idea(startup_idea, client)

    with open(
        "tavily_search_results.json", "w", encoding="utf-8"
    ) as file:
        json.dump(results, file, indent=2, ensure_ascii=False)

    print("\nResults saved to tavily_search_results.json")


if __name__ == "__main__":
    main()
```

```python

import os
import json
from tavily import TavilyClient
from tavily_search import search_startup_idea

# Create the Tavily client using your configured API key
api_key = os.getenv("TAVILY_API_KEY")

if not api_key:
    raise ValueError("TAVILY_API_KEY is missing. Run the API-key setup cell first.")

client = TavilyClient(api_key=api_key)

# Three startup ideas from different industries
test_ideas = [
    {
        "industry": "EdTech",
        "idea": "An app that gives rural students affordable AI tutoring in regional languages over WhatsApp."
    },
    {
        "industry": "HealthTech",
        "idea": "A subscription service that delivers personalised diabetic-friendly meal kits to urban households."
    },
    {
        "industry": "AgriTech",
        "idea": "A marketplace connecting small farmers directly with restaurants to reduce middlemen and food waste."
    }
]

all_test_results = {}

for item in test_ideas:
    industry = item["industry"]
    idea = item["idea"]

    print("\n" + "=" * 70)
    print("INDUSTRY:", industry)
    print("STARTUP IDEA:", idea)

    results = search_startup_idea(idea, client)

    all_test_results[industry] = {
        "startup_idea": idea,
        "results": results
    }

# Save all three tests in one JSON file
with open("milestone2_task4_test_results.json", "w", encoding="utf-8") as file:
    json.dump(all_test_results, file, indent=2, ensure_ascii=False)

print("\nAll three startup ideas have been tested.")
print("Results saved to milestone2_task4_test_results.json")
```

```python

import pandas as pd
import json

with open(
    "milestone2_task4_test_results.json",
    "r",
    encoding="utf-8"
) as file:
    test_results = json.load(file)

rows = []

for industry, data in test_results.items():
    for category, results in data["results"].items():
        for result in results:
            rows.append({
                "Industry": industry,
                "Startup Idea": data["startup_idea"],
                "Category": category,
                "Title": result.get("title", ""),
                "URL": result.get("url", ""),
                "Summary": result.get("content", ""),
                "Relevance": "",
                "Review Notes": ""
            })

df = pd.DataFrame(rows)

df.to_csv(
    "milestone2_task4_validation_report.csv",
    index=False,
    encoding="utf-8"
)

print("CSV validation report created!")
print("Total search results:", len(df))
display(df[["Industry", "Category", "Title", "URL"]])
```
